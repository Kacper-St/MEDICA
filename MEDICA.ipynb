```python
# Kółko i krzyżyk w Pythonie (wersja tekstowa)

def print_board(board):
    print("\n")
    for row in range(3):
        print(" | ".join(board[row * 3:(row + 1) * 3]))
        if row < 2:
            print("--+---+--")
    print("\n")


def check_winner(board, player):
    win_combinations = [
        [0,1,2], [3,4,5], [6,7,8],  # wiersze
        [0,3,6], [1,4,7], [2,5,8],  # kolumny
        [0,4,8], [2,4,6]            # przekątne
    ]

    for combo in win_combinations:
        if all(board[i] == player for i in combo):
            return True
    return False


def board_full(board):
    return all(space in ["X", "O"] for space in board)


def tic_tac_toe():
    board = [str(i + 1) for i in range(9)]
    current_player = "X"

    while True:
        print_board(board)

        move = input(f"Gracz {current_player}, wybierz pole (1-9): ")

        if not move.isdigit():
            print("Podaj liczbę!")
            continue

        move = int(move) - 1

        if move < 0 or move > 8:
            print("Pole poza zakresem!")
            continue

        if board[move] in ["X", "O"]:
            print("To pole jest już zajęte!")
            continue

        board[move] = current_player

        if check_winner(board, current_player):
            print_board(board)
            print(f"Gracz {current_player} wygrał!")
            break

        if board_full(board):
            print_board(board)
            print("Remis!")
            break

        current_player = "O" if current_player == "X" else "X"


# Uruchomienie gry
tic_tac_toe()
```
